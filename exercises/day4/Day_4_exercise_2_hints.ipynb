# Solutions

## Part 1 - Comments and docstrings

**1.** The function counts the `A` and `T` bases (not the `G`s!) and returns their **percentage** in the sequence. The comment is wrong.

````python
def f(s):
    # count the G's
    n = 0
    for x in s:
        if x == 'A' or x == 'T':
            n += 1
    return n / len(s) * 100

f('ATGCAT')
````

**2.**

````python
def at_content(sequence):
    """
    Calculate the AT content of a DNA sequence.

    Parameters:
        sequence: DNA sequence as a string, e.g. 'ATGCAT'

    Returns:
        The percentage of A and T bases in the sequence, as a float.
    """
    count = 0
    for base in sequence:
        if base == 'A' or base == 'T':
            count += 1
    return count / len(sequence) * 100

at_content('ATGCAT')
````

## Part 2 - Handling errors

**1.** The loop stops at `'N/A'` with:

```
ValueError: invalid literal for int() with base 10: 'N/A'
```

**2. and 3.**

````python
read_lengths = ['150', '148', '151', 'N/A', '150', 'error', '149']

valid_lengths = []

for length in read_lengths:
    try:
        number = int(length)
    except ValueError:
        print(f"Skipping value: {length}")
        continue
    valid_lengths.append(number)

average = sum(valid_lengths) / len(valid_lengths)
print(f"Average read length: {average}")
````

## Part 3 - User input

**2. and 3.**

````python
min_length = None

while min_length is None:
    answer = input("Give a minimum read length: ")
    try:
        min_length = int(answer)
    except ValueError:
        print(f"'{answer}' is not an integer. Try again!")

long_reads = []
for length in valid_lengths:
    if length >= min_length:
        long_reads.append(length)

print(f"Reads with length >= {min_length}: {long_reads}")
````
